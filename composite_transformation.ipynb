```python
from matplotlib import pyplot as plt
import math
```

```python
def translateTraingle(x_original, y_original, tx, ty):
    x_translated = []
    y_translated = []

    for x in x_original:
        x_translated.append(x + tx)

    for y in y_original:
        y_translated.append(y + ty)

    return x_translated, y_translated

def scaleTraingle(x_original, y_original, sx, sy):
    x_scaled = []
    y_scaled = []

    for x in x_original:
        x_scaled.append(x * sx)

    for y in y_original:
        y_scaled.append(y * sy)

    return x_scaled, y_scaled

def rotateTraingle(x_original, y_original, angle):
    x_rotated = []
    y_rotated = []

    for x ,y in zip(x_original, y_original):
        x_new = (x * math.cos(math.radians(angle))) - (y * math.sin(math.radians(angle)))
        y_new = (x * math.sin(math.radians(angle))) + (y * math.cos(math.radians(angle)))
        x_rotated.append(x_new)
        y_rotated.append(y_new)

    return x_rotated, y_rotated
```

```python
x_original = []
y_original = []

x_original.append(5)
x_original.append(10)
x_original.append(12)
x_original.append(5)

y_original.append(10)
y_original.append(15)
y_original.append(12)
y_original.append(10)

x_initial = x_original[:]
y_initial = y_original[:]

tx = 5
ty = 5
x_translated, y_translated = translateTraingle(x_original, y_original, tx, ty)

sx = 3
sy = 3
x_scaled, y_scaled = scaleTraingle(x_original, y_original, sx, sy)

angle = 75
x_rotated, y_rotated = rotateTraingle(x_original, y_original, angle)

plt.plot(x_initial,y_initial, marker='o',markersize=3,markerfacecolor='red')
# plt.plot(x_translated,y_translated, marker='o',markersize=3,markerfacecolor='green')
# plt.plot(x_scaled,y_scaled, marker='o',markersize=3,markerfacecolor='green')
plt.plot(x_rotated,y_rotated, marker='o',markersize=3,markerfacecolor='blue')

plt.show()
```
